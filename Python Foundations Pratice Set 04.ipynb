### **mini project on file handling in python**

```python
from pathlib import Path
import os

def read_file_folder():
    path = Path("")
    items = list(path.rglob("*.txt"))
    for i,items in enumerate(items):
        print(f"{i+1}. {items}")

def create_file():
    try:
        print("The Files in the current directory are:")
        read_file_folder()
        name = input("Enter the name of the file to be created: ")
        p = Path(name)
        if p.exists():
            print("File already exists")
        else:
             with open(p,"w") as fs:
                data = input("Enter the content of the file: ")
                fs.write(data+"\n")
             print(f"File {name} created successfully")
        print("The Files in the current directory are:")
        read_file_folder()

    except Exception as err:
        print(f"Error: {err}")

def read_file():
    try:
        print("The Files in the current directory are:")
        read_file_folder()
        name = input("Enter the name of the file to be read: ")
        p = Path(name)
        if p.exists() and p.is_file():
            print(f"Content of the file {name} is:\n")
            with open(p,"r") as fs:
                for line in fs:
                    print(line,end="")
        else:
            print("File not found or inaccessible.")
    except Exception as err:
        print(f"Error: {err}")

def update_file():
    try:
        print("The Files in the current directory are:")
        read_file_folder()
        name = input("Enter the name of the file to be updated: ")
        p = Path(name)
        if p.exists() and p.is_file():
            print("press 1 to append data to the file")
            print("press 2 to overwrite the file")
            print("press 3 to rename the file")
            res = int(input("Enter your choice: "))

            if res == 1:
               with open(p,"a") as fs:
                    data = input("Enter the content to be appended: ")
                    fs.write(data+"\n")
               print(f"Content appended to the file {name} successfully")

            if res == 2:
                with open(p,"w") as fs:
                    data = input("Enter the content to be written: ")
                    fs.write(data+"\n")
                print(f"File {name} updated successfully")
            if res == 3:
                new_name = input("Enter the new name of the file: ")
                if new_name and Path(new_name).exists() and Path(new_name).is_file():
                    print("A file with the same name already exists.")
                else:
                    p.rename(new_name)
                    print(f"File {name} renamed to {new_name} successfully")
        else:
            print("File not found or inaccessible.")

    except Exception as err:
        print(f"Error: {err}")

def delete_file():
    try:
        print("The Files in the current directory are:")
        read_file_folder()
        name = input("Enter the name of the file to be deleted: ")
        p = Path(name)
        if p.exists() and p.is_file():
           os.remove(p)
           print(f"File {name} deleted successfully")
        else:
            print("File not found or inaccessible.")
    except Exception as err:
        print(f"Error: {err}")

def main():
    print("\nWelcome to the File Handling Program")
    print("Press 1 to create a new file")
    print("Press 2 to read a file")
    print("Press 3 to update a file")
    print("Press 4 to delete a file")
    print("Press 5 to exit")


def choice():
    try:
        user_choice = int(input("Enter your choice: "))
        if user_choice not in [1, 2, 3, 4, 5]:
            print("Invalid choice. Please enter 1 to 5.")
            return None
        return user_choice
    except ValueError:
        print("Please enter a valid number.")
        return None


repeat = True

while repeat:
    main()
    user_choice = choice()

    if user_choice == 1:
        create_file()
    elif user_choice == 2:
        read_file()
    elif user_choice == 3:
        update_file()
    elif user_choice == 4:
        delete_file()
    elif user_choice == 5:
        repeat = False
        print("Exiting the program.")
        continue
    else:
        continue
    re = input("Do you want to continue? (y/n): ").lower()

    if re == "n":
        repeat = False
        print("Exiting the program.")
    elif re != "y":
        print("Invalid input. Please enter 'y' or 'n'.")
```

### **Inheritance in python**

#### **single level inheritance**

```python
class parent: # parent class / super class
    a = "THIS A CLASS VARIABLE OF PARENT CLASS"

class child(parent): # child class / subclass
    pass

Obj1 = parent()
Obj2 = child()
print(Obj1.a)
print(Obj2.a)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def show(self):
        print(f"Name: {self.name}")
class Dog(Animal):
    pass

human = Animal("Human")
dog = Dog("Dog")

# class Animal is the parent class and human is an object of the class Animal so it can access the methods and attributes of the class Animal.
# class Dog is the child class of the class Animal and dog is an object of the class Dog so it can access the methods and attributes of the class Animal.

human.show()
dog.show()
```

```python
class Animal: # parent class / super class
    def __init__(self,name):
        self.name = name
    def show(self):
        print(f"Name: {self.name}")

class Human(Animal):      # child class / subclass
    def __init__(self,name,age):
        super().__init__(name) # used when we want the same name attribute in the child class as in the parent class.
        self.age = age
    def show(self):
        print(f"Name: {self.name}, Age: {self.age}")

# here we are using the super() function to call the constructor of the parent class and initialize the name attribute in the child class.
# here we are doing method overriding(same function name different functionality) on show() method of the parent class in the child class to display the name and age of the human object.

dog = Animal("Dog")
human = Human("Human", 25)
dog.show()
human.show()
```

#### **MULTIPLE INHERITANCE**

```python
class mother:
    def __init__(self,name):
        self.name = name

class father:
    def __init__(self,age):
        self.age = age

class child(mother,father):
    def __init__(self,name,age):
        mother.__init__(self,name)
        father.__init__(self,age)

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}")

# follows the order of inheritance from left to right, so it will first look for the constructor in mother class and then in father class called as method resolution order(MRO)
child = child("Satwik", 25)
child.show()
```

#### **Multilevel inheritance**

```python
class grandparents:
    def __init__(self,name):
        self.name = name

class parents(grandparents):
    def __init__(self,name,age):
        super().__init__(name)
        self.age = age

class child(parents):
    def __init__(self,name,age,gender):
        super().__init__(name,age)
        self.gender = gender

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}")

child1 =  child("Satwik", 25, "Male")
child1.show()
```

### **Polymorphism in python**

```python
#Method overriding is an example of polymorphism in python where we have same method name in parent and child class but different functionality.
class parents:
    def show(self):
        print("This is a method of the parent class")

class children(parents):
    def show(self):
        print("This is a method of the child class")

child1 = children()
child1.show()

# once overriding is done in the child class, the method of the parent class is not accessible using the object of the child class.
```

### **Encapulsation in python**

```python
class bank_account:
    def __init__(self,name,age):
        self.name = name # public variable
        self._age = age # protected variable
        self.__balance = 0 # private variable

    def add_balance(self,amount):
        self.__balance += amount
        print(f"Balance added: {amount}, New balance: {self.__balance}")

    def withdraw_balance(self,amount):
        if amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Balance withdrawn: {amount}, New balance: {self.__balance}")

    def profile(self):
        print(f"Name: {self.name}, Age: {self._age}, Balance: {self.__balance}")


account_holder1 = bank_account("Satwik", 21)
account_holder1.profile()
account_holder1.add_balance(5000)
account_holder1.withdraw_balance(2000)
account_holder1.profile()

#Python creates a separate attribute named __balance on the object. It does not modify the class's name-mangled balance attribute.
account_holder1.__balance = 10000
print(account_holder1.__balance)
account_holder1.profile()
```

### **Abstraction in python**

```python
from abc import ABC, abstractmethod
class shape(ABC):   # abstract class
    @abstractmethod
    def area(self): # abstract method A method that is defined but not implemented in the abstract class. subclasses must provide the implementation.
        pass
    def perimeter(self):
        pass

class rectangle(shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = rectangle(5, 3)
print(f"Area: {rect.area()}")
print(f"Perimeter: {rect.perimeter()}")
```

```python

import json
import string
import random
from pathlib import Path


class Bank:

    database = "bank_data.json"
    data = []

    # Load existing account data
    try:
        if Path(database).exists():
            with open(database, "r") as fs:
                data = json.load(fs)

            if not isinstance(data, list):
                data = []
        else:
            data = []

    except (json.JSONDecodeError, OSError) as err:
        print(f"Error loading database: {err}")
        data = []

    # Save data to JSON file
    @classmethod
    def __update(cls):
        try:
            with open(cls.database, "w") as fs:
                json.dump(cls.data, fs, indent=4)
        except OSError as err:
            print(f"Error saving database: {err}")

    # Generate a unique account number
    @classmethod
    def __account_generation(cls):
        while True:
            alpha = random.choices(string.ascii_uppercase, k=5)
            digits = random.choices(string.digits, k=5)

            account_number = "".join(alpha + digits)

            if not any(
                account["account_number"] == account_number
                for account in cls.data
            ):
                return account_number

    # Authenticate an account
    @classmethod
    def __authenticate(cls):
        acc_num = input("Enter your account number: ").strip()
        pin = input("Enter your 4 digit PIN: ").strip()

        if not (pin.isdigit() and len(pin) == 4):
            print("Invalid PIN format.")
            return None

        for account in cls.data:
            if (
                account["account_number"] == acc_num
                and account["pin"] == pin
            ):
                return account

        print("Invalid account number or PIN.")
        return None

    # Create a new account
    def create_account(self):
        print("\n--- Create Account ---")

        name = input("Enter your name: ").strip()
        age_input = input("Enter your age: ").strip()
        mail_id = input("Enter your email ID: ").strip()
        phone_number = input("Enter your phone number: ").strip()
        pin = input("Create a 4 digit PIN: ").strip()

        # Validate age
        try:
            age = int(age_input)
        except ValueError:
            print("Age must be a valid number.")
            return

        if not 18 <= age <= 100:
            print("You must be between 18 and 100 years old.")
            return

        # Validate other details
        if not name:
            print("Name cannot be empty.")
            return

        if "@" not in mail_id or "." not in mail_id:
            print("Please enter a valid email address.")
            return

        if not (phone_number.isdigit() and len(phone_number) == 10):
            print("Phone number must contain exactly 10 digits.")
            return

        if not (pin.isdigit() and len(pin) == 4):
            print("PIN must contain exactly 4 digits.")
            return

        info = {
            "name": name,
            "age": age,
            "mail_id": mail_id,
            "phone_number": phone_number,
            "account_number": Bank.__account_generation(),
            "pin": pin,
            "balance": 0
        }

        Bank.data.append(info)
        Bank.__update()

        print("\nAccount created successfully!")
        print(f"Name: {info['name']}")
        print(f"Account Number: {info['account_number']}")
        print(f"Initial Balance: {info['balance']}")
        print("Please remember your account number and PIN.")

    # View account details
    def view_account(self):
        print("\n--- View Account ---")

        account = Bank.__authenticate()

        if account is None:
            return

        print("\nAccount Details")
        print("-------------------------")
        print(f"Name: {account['name']}")
        print(f"Age: {account['age']}")
        print(f"Email: {account['mail_id']}")
        print(f"Phone Number: {account['phone_number']}")
        print(f"Account Number: {account['account_number']}")
        print(f"Balance: {account['balance']}")
        print("-------------------------")

    # Deposit money
    def deposit_money(self):
        print("\n--- Deposit Money ---")

        account = Bank.__authenticate()

        if account is None:
            return

        amount_input = input("Enter the amount to deposit: ").strip()

        try:
            amount = int(amount_input)
        except ValueError:
            print("Please enter a valid whole-number amount.")
            return

        if not 0 < amount <= 100000:
            print("Deposit must be between 1 and 100000.")
            return

        account["balance"] += amount
        Bank.__update()

        print("Deposit successful!")
        print(f"Deposited Amount: {amount}")
        print(f"New Balance: {account['balance']}")

    # Withdraw money
    def withdraw_money(self):
        print("\n--- Withdraw Money ---")

        account = Bank.__authenticate()

        if account is None:
            return

        amount_input = input("Enter the amount to withdraw: ").strip()

        try:
            amount = int(amount_input)
        except ValueError:
            print("Please enter a valid whole-number amount.")
            return

        if amount <= 0:
            print("Withdrawal amount must be positive.")
            return

        if amount > account["balance"]:
            print("Insufficient balance.")
            return

        account["balance"] -= amount
        Bank.__update()

        print("Withdrawal successful!")
        print(f"Withdrawn Amount: {amount}")
        print(f"Remaining Balance: {account['balance']}")

    # Update account details
    def update_account(self):
        print("\n--- Update Account ---")

        account = Bank.__authenticate()

        if account is None:
            return

        while True:
            print("\nWhat would you like to update?")
            print("1. Name")
            print("2. PIN")
            print("3. Email")
            print("4. Phone Number")
            print("5. Return to Main Menu")

            try:
                option = int(input("Enter your choice: "))
            except ValueError:
                print("Please enter a valid number.")
                continue

            if option == 1:
                new_name = input("Enter your new name: ").strip()

                if new_name:
                    account["name"] = new_name
                else:
                    print("Name cannot be empty.")
                    continue

            elif option == 2:
                new_pin = input("Enter your new 4 digit PIN: ").strip()

                if new_pin.isdigit() and len(new_pin) == 4:
                    account["pin"] = new_pin
                else:
                    print("PIN must contain exactly 4 digits.")
                    continue

            elif option == 3:
                new_email = input("Enter your new email: ").strip()

                if "@" in new_email and "." in new_email:
                    account["mail_id"] = new_email
                else:
                    print("Please enter a valid email address.")
                    continue

            elif option == 4:
                new_phone = input("Enter your new phone number: ").strip()

                if new_phone.isdigit() and len(new_phone) == 10:
                    account["phone_number"] = new_phone
                else:
                    print("Phone number must contain exactly 10 digits.")
                    continue

            elif option == 5:
                return

            else:
                print("Invalid choice.")
                continue

            Bank.__update()
            print("Account details updated successfully!")

    # Delete an account
    def delete_account(self):
        print("\n--- Delete Account ---")

        account = Bank.__authenticate()

        if account is None:
            return

        confirmation = input(
            "Are you sure you want to delete your account? (y/n): "
        ).strip().lower()

        if confirmation == "y":
            Bank.data.remove(account)
            Bank.__update()
            print("Account deleted successfully.")

        elif confirmation == "n":
            print("Account deletion cancelled.")

        else:
            print("Invalid input. Please enter 'y' or 'n'.")


# Main program
def main():

    user = Bank()

    while True:
        print("\n================================")
        print("   BANK MANAGEMENT SYSTEM")
        print("================================")
        print("1. Create a New Account")
        print("2. View Account Details")
        print("3. Deposit Money")
        print("4. Withdraw Money")
        print("5. Update Account Details")
        print("6. Delete Account")
        print("7. Exit")
        print("================================")

        try:
            user_choice = int(input("Enter your choice: "))
        except ValueError:
            print("Please enter a valid menu number.")
            continue

        if user_choice == 1:
            user.create_account()

        elif user_choice == 2:
            user.view_account()

        elif user_choice == 3:
            user.deposit_money()

        elif user_choice == 4:
            user.withdraw_money()

        elif user_choice == 5:
            user.update_account()

        elif user_choice == 6:
            user.delete_account()

        elif user_choice == 7:
            print("Thank you for using our Bank Management System!")
            break

        else:
            print("Invalid choice. Please select between 1 and 7.")


if __name__ == "__main__":
    main()
```
