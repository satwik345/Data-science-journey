## **MATPLOTLIB**

### **LINE PLOT**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Create a basic line graph
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
# With customize color, linestyle and markers
plt.plot(x,y,color='r',linestyle='--',marker='o')
plt.show()
```

```python
# Change size of figure, marker,linewidth and markerfacecolor
plt.figure(figsize = (10,5))
plt.plot(x,y,color='r',linestyle='--',marker='o',linewidth=4,markersize=10,markerfacecolor='b')
plt.show()
```

```python
# Add title, xlabel, ylabel, gridlines
x = [1,2,3,4,5,6]
y = [5,6,7,8,9,10]
plt.figure(figsize = (10,5))
plt.plot(x,y,color='r',linestyle='--',marker='o',linewidth=4,markersize=10,markerfacecolor='b')
plt.xlabel('x axis -->')
plt.ylabel('y axis -->')
plt.title('X v/s Y graph')
plt.grid(True)
plt.show()
```

```python
y1 = [1,4,9,16,25]
y2 = [1,8,27,64,125]
x = [1,2,3,4,5]
plt.plot(x,y1,color='r',linestyle='--',marker='o',linewidth=4,markersize=10, markerfacecolor='b',label = 'Y1')
plt.plot(x,y2,color='b',linestyle='--',marker='o',linewidth=4,markersize=10, markerfacecolor='r',label = 'Y2')
plt.xlabel('x axis -->')
plt.ylabel('y axis -->')
plt.title('X v/s Y1,Y2 graph')
plt.legend()
plt.grid(True)
plt.show()
```

### **SCATTER PLOT**

```python
# Basic scatter plot
x = [1,2,3,4,5]
y = [10,20,25,30,40]
plt.scatter(x,y)
plt.xlabel('x axis -->')
plt.ylabel('y axis -->')
plt.title('scatter plot of x v/s y axis')
plt.grid(True)
plt.show()
```

```python
# change color,shape,size,transparency(alpha)
plt.scatter(x,y,color='r',marker='o',linewidth=7,alpha = 0.5,label = '1')
plt.xlabel('x axis -->')
plt.ylabel('y axis -->')
plt.title('scatter plot of x v/s y axis')
plt.grid(True)
plt.legend()
plt.show()
```

### **BAR CHART**

```python
# Create a vertical bar chart.
students = ["A","B","C","D"]
marks = [80,75,90,95]
plt.bar(students,marks)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students v/s Marks')
plt.grid(True)
plt.show()
plt.show()
```

```python
# Horizontal bar chart
students = ["A","B","C","D"]
marks = [80,75,90,95]
plt.barh(students,marks)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students v/s Marks')
plt.grid(True)
plt.show()
plt.show()
```

```python
# change color,width,alpha
plt.bar(students,marks,color='blue',width=0.4,label = 'Students')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students v/s Marks')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (10,5))
plt.bar(students,marks,color='blue',width=0.4,edgecolor='w',alpha = 0.5,linestyle='--',linewidth=2,label = 'Students')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Students v/s Marks')
plt.ylim(0,110)
plt.yticks(np.arange(0,120,20),fontsize=10)
plt.xticks(rotation=90)
plt.grid(axis='y')
plt.legend()
plt.show()
```

```python
# stacked bar chart
boys = [20,30,25]
girls = [10,20,15]
plt.bar(["A","B","C"],boys,label = 'Boys',linewidth = 2,edgecolor='w',alpha = 0.5,linestyle='--')
plt.bar(["A","B","C"],girls,bottom=boys,label = "Girls",color = 'pink',edgecolor='w',linewidth = 2,alpha = 0.5,linestyle='--')
plt.grid(axis='y')
plt.legend()
plt.show()
```

### **HISTOGRAM**

```python
data = np.random.randn(100)
plt.hist(data,bins=10,color='white',edgecolor='black',alpha = 0.5,density = True)
plt.xlabel('Ages')
plt.ylabel('Frequency')
plt.title('Histogram of Ages')
plt.grid(True)
plt.show()
```

```python
data = np.random.randn(100)
plt.hist(data,bins=10,edgecolor='w',density = True,histtype = 'step')
plt.xlabel('Ages')
plt.ylabel('Frequency')
plt.title('Histogram of Ages')
plt.grid(True)
plt.show()
```

### **PIE CHART**

```python
# basic pie chart
languages = ["Python", "SQL", "C++"]
students = [55, 25, 20]
plt.pie(students,labels=languages,colors=['blue','green','red'])
plt.title('Pie chart of students')
plt.grid(True)
plt.show()
```

```python
languages = ["Python", "SQL", "C++"]
students = [55, 25, 20]
plt.pie(students,colors=['blue','green','red'],autopct = '%1.1f%%',labels=['Python','SQL','C++'])
plt.title('Pie chart of students')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    students,
    labels=languages,
    colors=["gold","skyblue","lightgreen"],
    autopct='%1.1f%%',
    explode=(0.1,0,0),
    shadow=True,
    startangle=90,
    wedgeprops={'edgecolor':'black'},
    textprops={'fontsize':12}
)

plt.title("Programming Language Popularity")

plt.show()
```

### **SUBPLOTS**

```python
# Line Plot
x1 = [1, 2, 3, 4]
y = [10, 20, 30, 40]
z1 = [1,23,43,53]
fig,ax = plt.subplots()
ax.plot(x1,y,marker='o',linestyle='-',color='blue',markersize=10,label='x1')
ax.plot(z1 ,y,marker='o',linestyle='-',color='red',markersize=10,label='z1')
ax.set(xlabel='x axis -->',ylabel='y -->')
ax.set_title('Line graph')
plt.legend()
plt.show()
```

```python
# Bar Plot
subjects = ["Math", "Science", "English"]
marks = [90, 85, 88]
fig,ax = plt.subplots()
ax.bar(subjects,marks,color='blue',width=0.4,edgecolor='w',alpha = 0.5,linestyle='--',linewidth=2)
ax.set(xlabel='subjects -->',ylabel='marks -->')
ax.set_title('Bar chart of subjects vs marks')
plt.show()
```

```python
# Scatter plot
hours = [1, 2, 3, 4, 5]
marks = [30, 45, 60, 75, 90]
fig, ax = plt.subplots()
ax.scatter(hours, marks, color='blue', marker='o', label='hours', alpha = 0.9, linestyle='--', edgecolor='w', linewidth=1)
ax.set(xlabel='hours -->',ylabel='marks -->')
ax.set_title("Study Hours vs Marks")
ax.set_xlabel("Hours")
ax.set_ylabel("Marks")
plt.show()
```

```python
# Histogram
marks = [45, 50, 55, 60,45,56,78,35,53,54,56,65,46,43,57,97, 65, 70, 75, 80, 85, 90]
fig, ax = plt.subplots()
ax.hist(marks, bins=5,color='white',edgecolor='black',alpha = 0.5,density = True,linewidth = 2)
ax.set_title("Marks Distribution")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")
plt.show()
```

```python
# BOX PLOT
marks = [45, 50, 55, 60,45,56,78,35,53,54,56,65,46,43,57,97, 65, 70, 75, 80, 85, 90]
fig,ax = plt.subplots()
ax.boxplot(marks)
ax.set_title("Box Plot")
ax.set_xlabel("Box Plot")
ax.set_ylabel("Frequency")
plt.show()
```

```python
fig,ax = plt.subplots(2)
ax[0].hist(marks, bins=5)
ax[1].boxplot(marks)
plt.show()
```

## **SEABORN**

```python
import seaborn as sns
df = sns.load_dataset("titanic")
```

```python
sns.countplot(data=df,x="sex")
plt.show()
```

```python
sns.barplot(data=df,x='class',y='fare')
plt.show()
```

```python
sns.barplot(data=df,x='survived',y='fare')
plt.show()
```

```python
sns.boxplot(data=df, x="class", y="fare")
plt.show()
```

```python
sns.boxplot(data=df,x="class", y="age")
plt.show()
```

```python
sns.violinplot(data=df, x="class", y="age")
plt.show()
```

```python
sns.histplot(data=df, x="age",bins=10)
plt.show()
```

```python
g = sns.histplot(data=df, x="fare",hue="class",bins=10)
g.set_title("Histogram of fare")
g.set_xlabel("Fare")
g.set_ylabel("Frequency")
plt.show()
```

```python
g=sns.histplot(data=df, x="age", hue="sex",kde=True)
g.set(xlabel="Age",ylabel="Frequency")
g.set_title("Histogram of age")
plt.show()
```

```python
sns.kdeplot(data=df, x="age", hue="sex")
plt.show()
```

```python
sns.kdeplot(data=df, x="age", hue="survived")
plt.show()
```

```python
g=sns.scatterplot(data=df, x="age", y="fare",hue="survived",palette={1:"Green",0:"Red"},style="sex")
g.set(xlabel="Age",ylabel="Fare")
g.set_title("Scatter plot of age and fare")
plt.show()
```

```python
g=sns.scatterplot(data=df,x="age",y="fare",hue="class",size="fare")
g.set(xlabel="Age",ylabel="Fare")
g.set_title("Scatter plot of age and fare")
plt.show()
```

```python
sns.catplot(data=df,x="age",y="class",kind="bar",errorbar=None)
plt.show()
```

```python
sns.catplot(data=df,x="age",y="class",kind="box",showfliers=True,whis=[1,99])
plt.show()
```
