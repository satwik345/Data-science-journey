```python
matrix1 = np.array([[1, 2],
                    [3, 4]])
matrix2 = np.array([[5, 6],
                    [7, 8]])

# Add two matrices.
print(f" matrix1 + matrix2 = \n {matrix1 + matrix2}")
# Sub two matrices.
print(f" matrix1 - matrix2 = \n {matrix1 - matrix2}")
# Multiply two matrices.
print(f" matrix1 * matrix2 = \n {matrix1 * matrix2}")
# Divide two matrices.
print(f" matrix1 / matrix2 = \n {matrix1 / matrix2}")
# matrix multiplication
print(f" matrix1 x matrix2 = \n {np.matmul(matrix1, matrix2)}")
# Dot product
print(f"dot product = \n {np.dot(matrix1, matrix2)}")
# Transpose
print(f" transpose = \n {np.transpose(matrix1)}")
# determinant
print(f"determinant = {np.linalg.det(matrix1)}")
# inverse
print(f"inverse = \n {np.linalg.inv(matrix1)}")
# eigen values
print(f"eigen values = \n {np.linalg.eig(matrix1)}")
# trace
print(f" sum of all diagonal elements(trace)= {np.trace(matrix1)}")
#diagonal elements
print(f'diagonal elements = {np.diag(matrix1)}')
```

```python
# Adding columns
students['grade'] ='A'
students['pass/fail'] = 'pass'
employees['Bonus 5%'] = employees['salary'] * 0.05
orders['Total sales'] = orders['price'] * orders['quantity']

# Update values
students.loc[students['name']=='Satwik','marks'] = 92
students.loc[students['name']=='Riya','city'] = 'Kolkata'
students.loc[students['name'] == 'Satwik','age'] = 21
students.loc[students['name'] == 'Kiran','marks'] = 87

# Rename column names
students.rename(columns={'marks':'score'}, inplace=True)
students.rename(columns={'course':'Course','score':'Score','name':'Name','age':'Age','gender':'Gender','city':'City','grade':'Grade'}, inplace=True)

# Drop columns
employees.drop('Bonus 5%', axis=1, inplace=True)
```

```python
# apply method : Used when you want custom logic/function on data.
employees['bonus'] = employees['salary'].apply(lambda x:x*0.05)
students['pass/fail'] = students['Score'].apply(lambda x: "pass" if x>50 else "fail")

def grading(x):
    x = int(x)

    if 90 <= x <= 100:
        return "A"
    elif 80 <= x < 90:
        return "B"
    elif 70 <= x < 80:
        return "C"
    elif 60 <= x < 70:
        return "D"
    elif 50 <= x < 60:
        return "E"
    else:
        return "F"
students['Grade'] = students['Score'].apply(grading)
students['Score'].apply('sum')
```

```python
# Handling Missing Values
students.isnull()
students.isnull().any(axis=0)
students.isnull().sum()
students.fillna(students['Score'].mean())
students['City'].fillna('Unknown')
students.dropna()
```

```python
# Handle Duplicate values
students.duplicated()
students.drop_duplicates(subset= ['Name','City','Score'])
```

```python
# GroupBy
students.groupby('City')['Score'].mean().sort_values(ascending=False)
employees.groupby('department')['salary'].sum().sort_values(ascending=False)
students.groupby('City')['Name'].count().sort_values(ascending=False)
orders.groupby('category')['Total sales'].mean()
employees.groupby('department')['salary'].max().sort_values(ascending=False)
```

```python
# Pivot Table
employees.pivot_table(values = 'salary', index = 'department')
students.pivot_table(values='Score', index = 'Gender', aggfunc=['sum','mean','median'])
attendance.pivot_table(values = 'status', index = 'date', columns='status', aggfunc='count')
students.pivot_table(values='Name', index='Course', columns='City', aggfunc='count', fill_value=0)
orders.pivot_table(values = 'Total sales', index='city', columns='category', aggfunc='sum', fill_value=0, margins=True, margins_name='Total')
students.pivot_table(values='Score', index = 'Gender', columns='City', aggfunc=['count','sum'], margins=True, fill_value = 0, margins_name='Total')
```
