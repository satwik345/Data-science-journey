### **ADVANCE PANDAS**

```python
# Find the actor appearing most in Netflix content
actor = (
    df['cast']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

actor_df = actor.reset_index()
actor_df.columns = ["Actor", "Count"]

actor_df.head(1)
```

```python
# Find top genre in India
india_content = df[df['country'] == 'India']
top_genre_india = india_content['listed_in'].str.split(',').explode().value_counts().head(10)
top_genre_india_df = top_genre_india.reset_index()
top_genre_india_df.columns = ['Genre', 'Count']
top_genre_india_df.head(1)
```

```python
# Find the top 5 actors appearing in Indian movies.
top5_indian_actors = india_content['cast'].str.split(',').explode().str.strip().value_counts().head(5)
print(top5_indian_actors)
```

```python
current_year = 2026

df['content_age'] = current_year - df['release_year']

print(df[['title', 'release_year', 'content_age']].head())
```

```python
# Categorize content
def category(year):
    if year < 2000:
        return 'Old'
    elif year <= 2015:
        return 'Mid'
    else:
        return 'New'

df['Category'] = df['release_year'].apply(category)
```

```python
movies = df[df['type'] == 'Movie'].dropna().copy()

movies['duration'] = (
    movies['duration']
    .str.replace(' min', '', regex=False)
    .astype(int)
)

correlation = movies['release_year'].corr(movies['duration'])

print(correlation)
```

```python
# Find the top 5 genres released after 2020.
after2020 = df[df['release_year'] > 2020]
top5genres_after2020 = after2020['listed_in'].str.split(',').explode().str.strip().value_counts().head()
print(top5genres_after2020)
```

```python
# Find the director with the highest number of Comedy movies.
director_counts = (
    netflix_comedy_movies
    .dropna(subset=['director'])
    .groupby('director')['show_id']
    .count()
    .sort_values(ascending=False)
)

print(director_counts.head(1))
```
