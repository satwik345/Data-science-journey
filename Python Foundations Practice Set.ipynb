```python
# Function to calculate factorial.
def factorial(x):
    fact = 1
    if x == 1 or x == 0:
        print(f"factorial: {fact}.")
    else:
        for i in range(1,x+1):
            fact *= i
        print(f"factorial: {fact}.")
factorial(5)

def factorial_recursive(x):
    if x == 0 or x == 1:
        return 1
    # Recursive case: x * (x - 1)!
    else:
        return x * factorial_recursive(x - 1)
print(f"using recursive function: {factorial_recursive(5)}")
```

```python
students = [
    {"id":101,"name":"Satwik","marks":85},
    {"id":102,"name":"Rahul","marks":72},
    {"id":103,"name":"Anjali","marks":91}
]

def menu():
    print("""
CHOSE FROM OPTIONS BELOW:

1.show
2.add
3.remove
4.update
5.search
6.High performing students
7.low performing students
8.Class Statistics
9.exit program
--------------------------------------------------------------------------------------""")

def confirmation():
    try:
        final = input("Are you sure? (yes/no): ").lower()
        if final == "yes":
            return True

        elif final == "no":
            print("Exiting program...")
            return False

        else:
            raise ValueError("Invalid option")

    except ValueError as e:
        print("Error:", e)
        return False

def show_students():
    print("-" * 40)
    print("ID\tName\tMarks")
    print("-" * 40)
    for student in students:
        print(f'{student["id"]}\t{student["name"]}\t{student["marks"]}')

def add_student():
    try:
        new_student_id = int(input("Enter student ID: "))
        for student in students:
            if student.get("id") == new_student_id:
                raise ValueError("Student already exists")

        new_student_name = input("Enter student name: ").title()
        if not new_student_name.replace(" ", "").isalpha():
            raise ValueError("Student name must contain only alphabets")

        new_student_marks = float(input("Enter student marks: "))
        if new_student_marks < 0 or new_student_marks > 100:
            raise ValueError("Marks must be between 0 and 100")

        if confirmation():
            students.append({ "id": new_student_id, "name": new_student_name, "marks": new_student_marks})
            print("Student added successfully")
        else:
            print("Cancelled")
            return

    except ValueError as e:
        print("Error:", e)

def remove_student():
    try:
        basis = input("What basis to remove (id/name)? ").lower()

        if basis == "id":
            student_id = int(input("Enter student ID: "))
            for student in students:
                if student["id"] == student_id:
                    if confirmation():
                        students.remove(student)
                        print("Removed successfully")
                        return
                    else:
                        print("Deletion cancelled")
                        return

        elif basis == "name":
            student_name = input("Enter student name: ").title()
            for student in students:
                if student["name"] == student_name:
                    if confirmation():
                         students.remove(student)
                         print("Removed successfully")
                         return
                    else:
                         print("Deletion cancelled")
                         return

        else:
            raise ValueError("Invalid basis")

        print("Student not found")

    except ValueError as e:
        print("Error:", e)

def updating_students():
    try:
        update_basis = input("What do you want to update (id/name/marks)? ").lower()

        if update_basis == "id":
            current_student_id = int(input("Enter current student ID: "))
            further_student_id = int(input("Enter new student ID: "))
            for student in students:
                if student["id"] == further_student_id:
                    raise ValueError("Student ID already exists")
            for student in students:
                if student["id"] == current_student_id:
                    if confirmation():
                        student["id"] = further_student_id
                        print("Student ID updated")
                        return
                    else:
                        print("Cancelled")
                        return

        elif update_basis == "name":
            update_name_id = int(input("Enter student ID: "))
            update_name = input("Enter new student name: ").title()
            if not update_name.replace(" ", "").isalpha():
                raise ValueError("Student name must contain only alphabets")
            for student in students:
                if student["id"] == update_name_id:
                    if confirmation():
                        student["name"] = update_name
                        print("Student name updated")
                        return
                    else:
                        print("Cancelled")
                        return

        elif update_basis == "marks":
            update_marks_id = int(input("Enter student ID: "))
            update_marks = float(input("Enter new marks: "))
            if update_marks < 0 or update_marks > 100:
                raise ValueError("Marks must be between 0 and 100")
            for student in students:
                if student["id"] == update_marks_id:
                    if confirmation():
                        student["marks"] = update_marks
                        print("Student marks updated")
                        return
                    else:
                        print("Cancelled")
                        return


        else:
            raise ValueError("Invalid basis")
        print("Student not found")

    except ValueError as e:
        print("Error:", e)

def search_students():
    try:
        if not students:
            print("No students available")
            return
        search_basis = input("What do you want to search (id/name)? ").lower()
        if search_basis == "id":
            search_id = int(input("Enter student ID: "))
            for student in students:
                if student["id"] == search_id:
                        print("\nStudent Found:")
                        print(f'ID: {student["id"]}, Name: {student["name"]}, Marks: {student["marks"]}')
                        return

        elif search_basis == "name":
                search_name = input("Enter student name: ").title()
                for student in students:
                    if student["name"] == search_name:
                        print("\nStudent Found:")
                        print(f'ID: {student["id"]}, Name: {student["name"]}, Marks: {student["marks"]}')
                        return
        else:
            raise ValueError("Invalid basis")
        print("Student not found")

    except ValueError as e:
        print("Error:", e)

def top_10_students():
    if not students:
        print("No students available")
        return
    top_students = sorted(students, key=lambda x: x["marks"], reverse=True)[:10]
    print("\nTop 10 Students")
    print("-" * 30)
    for student in top_students:
        print(f'ID: {student["id"]}, Name: {student["name"]}, Marks: {student["marks"]}')

def last_10_students():
    if not students:
       print("No students available")
       return

    last_students = sorted(students, key=lambda x: x["marks"])[:10]
    print("\nLast 10 Students")
    print("-" * 30)
    for student in last_students:
        print(f'ID: {student["id"]}, Name: {student["name"]}, Marks: {student["marks"]}')

def students_statistics():
    if not students:
          print("No students available")
          return

    highest = max(student["marks"] for student in students)
    lowest = min(student["marks"] for student in students)
    average = sum(student["marks"] for student in students) / len(students)

    passed = sum(1 for student in students if student["marks"] >= 40)
    failed = len(students) - passed

    print("\nClass Statistics")
    print("-" * 40)
    print(f"Highest Marks: {highest}")
    print(f"Lowest Marks: {lowest}")
    print(f"Average Marks: {average:.2f}")
    print(f"Passed Students: {passed}")
    print(f"Failed Students: {failed}")

while True:
    menu()
    try:
        choice = int(input("Enter an option: "))
        if choice == 1:
            show_students()
        elif choice == 2:
            add_student()
        elif choice == 3:
            remove_student()
        elif choice == 4:
            updating_students()
        elif choice == 5:
            search_students()
        elif choice == 6:
            top_10_students()
        elif choice == 7:
            last_10_students()
        elif choice == 8:
            students_statistics()
        elif choice == 9:
            print("Exiting...")
            break
        else:
            raise ValueError("Invalid option")
    except ValueError as e:
        print("Error:", e)
```

#### **PERSONAL EXPENSE TRACKER**

```python
expenses = [
    {"category":"Food","amount":500},
    {"category":"Travel","amount":800}
]

def expenses_menu():
    print("""
CHOOSE FROM THE GIVEN OPTION:

1. SHOW EXPENSES
2. ADD EXPENSES
3. REMOVE EXPENSES
4. UPDATE EXPENSES
5. SEARCH EXPENSES
6. EXPENSES ANALYTICS
7. EXIT MENU
    """)

def confirmation():
    while True:
        final = input("ARE YOU SURE? (y/n): ").lower()

        if final == "y":
            return True
        elif final == "n":
            return False
        else:
            print("Invalid option")

def empty_list_check():
    if len(expenses) == 0:
        print("No expenses available")
        return False
    return True

def show_expenses():
    if not empty_list_check():
        return
    print("-" * 30)
    print("Category\tAmount")
    print("-" * 30)
    for i in expenses:
        print(f'{i["category"]}\t{i["amount"]}')

def add_expense():
    try:
        new_category = input("Enter new category: ").title()
        if not new_category.replace(" ", "").isalpha():
            raise ValueError("Invalid category")
        new_amount = float(input("Enter new amount: "))
        if new_amount <= 0:
            raise ValueError("Invalid amount")
        for i in expenses:
            if i["category"] == new_category:
                raise ValueError("Category already exists")
        if confirmation():
           expenses.append({"category": new_category, "amount": new_amount})
           print("Added new expense")
        else:
           print("Cancelled.")

    except ValueError as e:
        print("Error:", e)

def remove_expense():
    try:
        if empty_list_check():
            remove_category = input("Enter category to remove: ").title()
            for i in expenses:
                if i["category"] == remove_category:
                    if confirmation():
                        expenses.remove(i)
                        print("Removed expense")
                        return
                    else:
                        print("Cancelled.")
                        return
            print("Expense not found")
        else:
            print("EMPTY LIST")

    except ValueError as e:
        print("Error:", e)

def update_expense():
    try:
        if not empty_list_check():
            return

        update_field = input("Update Category or Amount? ").lower()
        if update_field not in ["category", "amount"]:
            raise ValueError("Invalid field")
        category_name = input("Enter expense category: ").title()
        if not category_name.replace(" ", "").isalpha():
            raise ValueError("Invalid category")

        for i in expenses:
            if i["category"] == category_name:
                if update_field == "category":
                    new_name = input("Enter new category name: ").title()
                    if not new_name.replace(" ", "").isalpha():
                        raise ValueError("Invalid category")
                    if confirmation():
                        i["category"] = new_name
                        print("Updated successfully")
                        return
                elif update_field == "amount":
                    new_amount = float(input("Enter new amount: "))
                    if new_amount <= 0:
                        raise ValueError("Invalid amount")
                    if confirmation():
                        i["amount"] = new_amount
                        print("Updated successfully")
                        return
                else:
                    raise ValueError("Invalid field")

        print("Expense not found")

    except ValueError as e:
        print("Error:", e)

def search_expenses():
    try:
        if not empty_list_check():
            return
        search_expense = input("Enter search expense: ").title()
        if not search_expense.replace(" ", "").isalpha():
            raise ValueError("Invalid category")
        for i in expenses:
            if i["category"] == search_expense:
                print(f"{i['category']}\t{i['amount']}")
                return

        print("Expense not found")

    except ValueError as e:
        print("Error:", e)

def expenses_analysis():

    if not empty_list_check():
            return

    highest_category = max(expenses, key=lambda x: x["amount"])

    lowest_category = min(expenses, key=lambda x: x["amount"])

    total_expense = sum(i["amount"] for i in expenses)

    total_categories = len(expenses)

    average_expense = total_expense / total_categories

    try:
         total_budget = float(input("Enter total budget: "))
         if total_budget <= 0:
            raise ValueError("Invalid budget")
    except ValueError as e:
        print("Error:", e)
        return

    left_budget = total_budget - total_expense

    sorted_expenses = sorted(expenses, key=lambda x: x["amount"], reverse=True)

    savings_rate = (left_budget / total_budget) * 100

    percentage_categories_wise = {}
    for category in sorted_expenses:
        percentage_categories_wise[category["category"]] = (category["amount"] / total_expense) * 100

    print("EXPENSES ANALYSIS:")
    print("-"*40)
    print("Total budget: ", total_budget)
    print("Total expense: ", total_expense)
    print("Total categories: ", total_categories)
    print("Average expense: ", average_expense)
    print(f'Savings rate: {savings_rate}%')
    print(f'Highest expense: {highest_category["category"]} - {highest_category["amount"]}')
    print(f'Lowest expense: {lowest_category["category"]} - {lowest_category["amount"]}')
    print("")
    print(F'FINANCIAL SITUATION ANALYSIS:')
    print("Total budget left: ", left_budget)
    if left_budget < 0:
        debt = abs(left_budget)
        print(f"Debt: {debt}")
    elif left_budget == 0:
        print("Budget finished")
    elif left_budget <= total_budget * 0.5:
        print("Half budget left")
    else:
        print("Good savings")
    print('')
    print("ALL EXPENSES SORTED HIGH-LOW:")
    for expense in sorted_expenses:
        print(f'{expense["category"]}: {expense["amount"]}')
    print('')
    print("EXPENSES V/S TOTAL PERCENTAGE:")
    for category, percentage in percentage_categories_wise.items():
        print(f"{category}\t{percentage:.2f}%")

while True:
    expenses_menu()
    try:
        choice = int(input("Enter an option: "))
        if choice == 1:
            show_expenses()
        elif choice == 2:
            add_expense()
        elif choice == 3:
            remove_expense()
        elif choice == 4:
            update_expense()
        elif choice == 5:
            search_expenses()
        elif choice == 6:
            expenses_analysis()
        elif choice == 7:
            print("Exiting...")
            break
        else:
            raise ValueError("Invalid option")
    except ValueError as e:
        print("Error:", e)
```
