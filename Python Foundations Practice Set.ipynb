```python
# Rotate list by one position.
original = [1,2,3,4]

left_shift = []
for i in original:
    if i == original[-1]:
        left_shift.append(i)
        break
for num in original:
    if num not in left_shift:
        left_shift.append(num)

right_shift = []
for i in original:
    if i == original[0]:
        continue
    right_shift.append(i)
for num in original:
    if num not in right_shift:
        right_shift.append(num)

print(f"RIGHT SHIFT BY ONE :{right_shift}")
print(f"LEFT SHIFT BY ONE :{left_shift}")
```

```python
# Function to find largest among 3 numbers.
tup = (1,2,3)
num1,num2,num3 = tup
def compare(num1,num2,num3):
    if num1 > num2 and num1 > num3:
        print(f"{num1} is largest.")
    elif num2 > num1 and num2 > num3:
        print(f"{num2} is largest.")
    elif num3 > num1 and num3 > num2:
        print(f"{num3} is largest.")
    else:
        print("Cant find")
compare(num1,num2,num3)
```

```python
# Function to calculate factorial.
def factorial(x):
    fact = 1
    if x == 1 or x == 0:
        print(f"factorial: {fact}.")
    else:
        for i in range(1,x+1):
            fact *= i
        print(f"factorial: {fact}.")
factorial(5)

def factorial_recursive(x):
    if x == 0 or x == 1:
        return 1
    # Recursive case: x * (x - 1)!
    else:
        return x * factorial_recursive(x - 1)
print(f"using recursive function: {factorial_recursive(5)}")
```

```python
# Function to check prime number.
def prime_or_not(x):
    if x <= 1:
        print(f"{x} is not Prime")
        return
    """only need to check up to square root of the number:"""
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            print(f"{x} is not Prime")
            break
    else:
        print(f"{x} is Prime")
prime_or_not(21)
```

```python
# Build a Restaurant Billing System Use: *args → food prices **kwargs → customer details
# Output: total bill, GST, final bill
def billing(*args, **kwargs):
    total_bill = sum(args)
    GST = total_bill * 0.05
    final_bill = total_bill + GST

    print(f"BILLING NO: {kwargs.get("bill_no")}")
    print(f"CUSTOMER NAME: {kwargs.get("name")}")
    print(F"TOTAL BILL: {total_bill}/-")
    print(F"GST: {GST}/-")
    print(F"FINAL BILL: {final_bill}/-")

billing(105,230,30,40,1515,5151,name="Satwik",bill_no=5145155)
```

```python
#Inventory management
inventory = {"apple": 5, "banana": 2}
while True:
    print("""
PRESS
1. Show inventory stock
2. Search stock
3. Add new stock
4. Remove stock
5. Update value of stock)
6. Exit
    """)

    option = input("Enter an option: ")

    if option == "1":
        print(inventory)

    elif option == "2":
        search = input("Enter stock name: ").lower()
        if search in inventory:
            print(f"{search}: {inventory[search]}")
        else:
            print("Stock not found")

    elif option == "3":
        key = input("Enter stock name: ").lower()
        value = int(input("Enter stock quantity: "))
        if key not in inventory:
            inventory[key] = value
        else:
            inventory[key] += value
        print("Stock added successfully")

    elif option == "4":
        delete_option = input("Fully remove or selective items (F/S): ").lower()
        if delete_option == "s":
            delete_key = input("Enter stock name: ").lower()
            if delete_key in inventory:
                delete_value = int(input("Enter quantity to remove: "))
                del inventory[delete_key]
                print("Stock updated")
            else:
                print("Stock not found")
        elif delete_option == "f":
            inventory.clear()
            print("Inventory cleared")
        else:
            print("Invalid choice")

    elif option == "5":
        update_key = input("Enter stock name: ").lower()
        if update_key in inventory:
            update_value = int(input("Enter new quantity: "))
            inventory[update_key] = update_value
            print("Inventory updated")
        else:
            print("Stock not found")

    elif option == "6":
        print("Exiting program...")
        break

    else:
        print("Invalid option")
```
